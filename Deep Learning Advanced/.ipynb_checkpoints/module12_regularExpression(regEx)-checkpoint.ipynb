# Character Set

```python
import re
```

```python
txt="The rain Spain"
x=re.findall(r"[a-b]",txt)
print(x)
```

# Escape Sequence

```python
txt = " Hi This is Dibya, My age is 25"
x = re.findall(r"\d+",txt) #d= digit, D=without digit all select,s=space select, S=without space all select
print(x)
```

# Digit character class

```python
txt = "that will be 59 dollars"
x = re.findall(r"\d",txt)
print(x)
```

```python
txt = "that will be 59 dollars"
x = re.findall(r"\d+",txt)
print(x)
```

```python
#match 2 ,  3 digits

txt = "that will be 596 56 dollars"
x = re.findall(r"\d{2}",txt)
print(x)
```

```python
#match 2 or 3 digits
txt = "that will be 592 43 1 dollars"
x = re.findall(r"\d{2,3}",txt)
print(x)
```

```python
txt = "this is my number 01905-672239"
x = re.findall(r"\d{5}-\d{6}",txt)
print(x)
```

```python
#at the beginning

txt = "123  139 this is my number 01905-672239"
x = re.findall(r"^\d+",txt)

print(x)
```

```python
# at the end


txt = "123  123 this is my number 01905-672239"
x = re.findall(r"\d+$",txt)

print(x)
```

# Wildcard character .

```python
txt = "hello"
x = re.findall(r".",txt)
print(x)
```

```python
txt = "hello"
x = re.findall(r"he..o",txt)
print(x)
```

# Asterics

```python
txt = "dibya@gmail.com"
x = re.findall(r".*@.*\..*",txt)# back slash er por j dot oita special character(.com er dot).duita k alada korar jonno back slash diye use kora. 

print(x)
```

```python
txt = "dibya@gmail.com  bat@gmail.com cat@gmail.com mat@gmail.com dog@yahoo.com"
x = re.findall(r".*@.*\.com$",txt)
print(x)
```

# Plus Quantifier

```python
txt = "heo"
x = re.findall(r"he.*o",txt)
print(x)
```

```python
txt = "heo"
x = re.findall(r"he.+o",txt) # here nothing to select
print(x)
```

```python
txt = "heo 12"
x = re.findall(r"\d+",txt)
print(x)
```

# Question mark

```python
txt = "hello helo"

x = re.findall(r"hel?o",txt) #jekhane l - 0 or 1 baaer moddhe thakbe sheta print hobe

print(x)
```

```python
txt = "heo hello he1o"

x = re.findall(r"he.?o",txt) #he diye start and o diye end.. (.?) here majhkhane 0or 1 occurance thakbe

print(x)
```

# Capture and group

```python
txt = "ababab"
x = re.findall("ab+",txt) # + means one or more occurance 
print(x)
```

```python
#email

txt = "abc@gmail.com"

x = re.findall(r"(\w+)@(\w+)\.com",txt) # \w+ means one or more character

print(x)
 
for first,second in x :
  print(first,second)
```

```python
#name ,age

txt = "John 25, Alice 30, Bob 18"

x =re.findall(r"(\w+)\s(\d+)",txt)
print(x)

for name,age in x:
  print(name,age)
```

# Functions


## findall

```python
txt = "My lucky numbers are 7,25,99"

x = re.findall(r"\d+",txt)
print(x)
```

# search

```python
txt = "My lucky number is 25 19 22 1"
x = re.search(r"\d+",txt)
print(x)

print(x.start())
print(x.end())
print(x.group())
```

```python
# email
txt = "contact us at support@phitron.io , you can also email us admin@yahoo.com"

match = re.search(r"\w+@\w+\.\w+",txt)

print(match.group())
```

## split

```python
txt = "apple, banana, orange"

x = re.split(r",",txt)
print(x)
```

## sub

```python
txt = "I have 2 cats and 3 dogs"
x = re.sub(r"\d+","#",txt) #replace kore

print(x)
```
