```python
from PIL import Image
import matplotlib.pyplot as plt

image = Image.open('beautiful-flowers-dahlia.webp')

plt.imshow(image)
# plt.axis('off')

```

```python
print(type(image))
```

```python
import numpy as np
image_array = np.array(image)
print(image_array)
```

```python
# Convert back to image
new_image = Image.fromarray(image_array)

# Show image
new_image.show()
new_image
```
