# Tokenize


```python
import re # regular expression

def tokenize(text):

  #1  Lowercase
  text = text.lower()

  #2 remove quotes
  text = re.sub(r"[\"']","",text)     # here, [\"'] # double quote and single quote remove. and replace with empty space


  #3 remove all punctuations
  text = re.sub(r"[^a-z0-9\s]", " ", text) # a to z alphabate and 0 to 9 all number thakbe.bakigula(punctuation) remove with empty space

  #4 remove extra space
  text = re.sub(r"\s+", " ", text).strip() # \s+ extra spae remove kore

  #tokenize split into words
  tokens = text.split()

  return tokens
```

```python
print(tokenize(df['question'][0]))
```

# Vocab forming

```python
vocab = {'<UNK>':0}
```

```python
def build_vocab(row):
  # print(row['question'],row['answer'])

  tokenized_question = tokenize(row['question'])
  tokenized_answer = tokenize(row['answer'])

  # print(tokenized_question,tokenized_answer)

  merged_tokens = tokenized_question + tokenized_answer

  # print(merged_tokens)

  for token in merged_tokens:
    if token not in vocab:
      vocab[token] = len(vocab)
```

```python
df.apply(build_vocab,axis = 1 ) # row paite chai tai axis=1
```

# Text to indices

```python
#index 
def text_to_indices(text,vocab):
  indexed_text = []

  for token in tokenize(text):
    # print(token)
    if token in vocab:
      # print(vocab[token])
      indexed_text.append(vocab[token])

    else:
      indexed_text.append(vocab['<UNK>'])

  return indexed_text
```

```python
text_to_indices("the is Phitron",vocab)
```

# Dataset and Dataloader

```python
import torch
from torch.utils.data import Dataset,DataLoader
```

```python
df.shape[0]
```

```python
index = 0

text_to_indices(df.iloc[index]['question'],vocab)
```

```python
class QADataset(Dataset):

  def __init__(self,df,vocab):
    self.df = df
    self.vocab = vocab

  def __len__(self):
    return self.df.shape[0]

  def __getitem__(self,index):
    numerical_question = text_to_indices(self.df.iloc[index]['question'],self.vocab)
    numerical_answer = text_to_indices(self.df.iloc[index]['answer'],self.vocab)

    return torch.tensor(numerical_question), torch.tensor(numerical_answer)
```

```python
dataset = QADataset(df,vocab)
```

```python
dataset[1]
```

```python
dataloader = DataLoader(dataset,batch_size=1,shuffle=True)
```

```python
for question,answer in dataloader:
  
  #print(question,answer)
  print(question,answer[0])
```

# RNN Architecture Implementation

```python
import torch.nn as nn

class simpleRNN(nn.Module):
  def __init__(self,vocab_size,embedding_dim=50,hidden_size=64): # embedding_dim 50 means per word 50 ath number nibe. hidden_size 64 means 64 neurons
    super().__init__()

    self.embedding = nn.Embedding(vocab_size,embedding_dim)
    self.rnn = nn.RNN(embedding_dim,hidden_size,batch_first=True) #batch_first true korle batch aage(1st ah) kaaj kore
    self.fc = nn.Linear(hidden_size,vocab_size)


  def forward(self,question):
     embedded = self.embedding(question) # shape= (1, seq_len, 50)
     _,final = self.rnn(embedded)

     return self.fc(final.squeeze(0))  # (1, vocab_size)

```

```python
model = simpleRNN(vocab_size = len(vocab))
```

```python
# EXample of unsqueeze

import torch

x = torch.tensor([1, 2, 3])   # shape: (3,)
print(x.shape)

y = x.unsqueeze(0)
print(y.shape)
```

# Training loop

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr = 0.0001)
epochs = 50
```

```python
for epoch in range(epochs):
    total_loss = 0
    for question, answer in dataloader:
        optimizer.zero_grad()
        output = model(question)
        
        # Fix: take only the first answer token as target → shape (1,)
        #print(answer)
        target = answer[0][0].unsqueeze(0) #squeeze and unsqeeze mainly shape tah milanor jonno kora hoy
        loss = criterion(output, target)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    if (epoch + 1) % 10 == 0:
        print(f"Epoch {epoch+1:3d}/{epochs}  |  Loss: {total_loss:.4f}")
```

# Prediction

```python
vocab.items()
```

```python
idx_to_word = { idx : word for word,idx in vocab.items()  }
idx_to_word
```

```python
def predict(question_text,model,vocab,idx_to_word):
  model.eval()

  with torch.no_grad():
    indices = text_to_indices(question_text,vocab)

    if not indices:
      return "<UNK>"

    x = torch.tensor(indices).unsqueeze(0)  #batch size, sequence length

    pred = torch.argmax(model(x),dim=1).item()


  return idx_to_word.get(pred,"<UNK>") # give pred or unknown
```

```python
# Test
tests = [
    "What is the capital of France?",
    "What is H2O commonly called?",
    "Which planet is known as the red planet?",
    "Which bird cannot fly?",
    "What is the capital of Japan ?",

]
for q in tests:
    print(f"{q:45} → {predict(q, model, vocab, idx_to_word)}")
```
