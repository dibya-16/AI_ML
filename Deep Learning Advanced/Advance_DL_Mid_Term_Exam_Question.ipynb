# Write Answer 01:
A CNN (Convolutional Neural Network) is a deep learning model mainly used for image processing.

1. Local connectivity: Filters examine small parts of an image to detect features like edges, eyes, and wrinkles.

## Explain padding and stride in CNNs in detail. Then explain the convolutional output size formula and what each symbol in the formula means. You do not need to derive the formula. Show how the output height and width change when you use valid padding, same padding, or a larger stride. Also explain why padding is useful when working with cropped face images.


# Write Answer 05:
Padding and Stride:

Padding: Adds extra pixels around the image. Valid padding makes the output smaller, while same padding keeps the output size similar to the input.
Stride: Determines how many pixels the filter moves at each step. A larger stride produces a smaller output.

Output size: Depends on the input size, kernel size, padding, and stride.

Example: A 64×64 image can become 62×62 with valid padding, 64×64 with same padding, or smaller with a larger stride.
For face images: Padding helps preserve important facial features near the edges of cropped images.

## Explain the concept of pooling in CNNs. Compare MaxPooling and AveragePooling in terms of operation, advantages, and impact on feature extraction. Discuss how pooling helps in reducing overfitting and computational complexity when training an age-detection CNN on thousands of face images.


# Write Answer 06:
Pooling reduces the size of feature maps while keeping important information.

MaxPooling: Selects the maximum value from a region. It keeps the strongest features, such as edges or wrinkles.

AveragePooling: Calculates the average value of a region. It provides smoother and more general feature information.

Advantages: Pooling reduces the number of parameters and computation, making training faster. It also reduces the risk of overfitting by simplifying feature maps. In age detection, it helps the CNN focus on important facial features while ignoring small variations.

# Question 07: [ Marks 10 ]

## Describe the convolution operation in CNNs with a complete step-by-step numerical example. Use a small input matrix (for example 5×5) and a 3×3 filter, apply a chosen stride and padding, and show how each value of the output feature map is calculated. Also compute the output size using the standard formula.



# Write Answer 07:
Suppose we have a 5×5 input image and a 3×3 filter, with stride = 1 and no padding.

Place the 3×3 filter on the top-left part of the image.

Multiply each filter value with the corresponding image value and add all the results. This gives the first output value.

Move the filter one step to the right and repeat the process.

Continue across the row and then move down until the whole image is covered.

The calculated values form the feature map.

For a 5×5 input, 3×3 filter, stride 1, and no padding:

Output size = (5 − 3) / 1 + 1 = 3×3

So, the convolution produces a 3×3 feature map.

````python

## Part A: Load the data

import os
import glob
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import Dataset, DataLoader, random_split
from torchvision import transforms

import kagglehub
````

````python
path = kagglehub.dataset_download(
    "arashnic/faces-age-detection-dataset"
)

print("Dataset path:", path)
````

````python
csv_files = glob.glob(
    os.path.join(path, "**", "*.csv"),
    recursive=True
)

print("CSV files found:")

for file in csv_files:
    print(file)
````

````python
CSV_PATH = csv_files[0]

print("Using CSV:", CSV_PATH)
````

````python
data = pd.read_csv(CSV_PATH)

print("Shape:", data.shape)
print()
print(data.head())
print()
print("Columns:", data.columns.tolist())
````

````python
image_files = []

for extension in ["*.jpg", "*.jpeg", "*.png"]:
    image_files.extend(
        glob.glob(
            os.path.join(path, "**", extension),
            recursive=True
        )
    )

print("Total images found:", len(image_files))
print()

for img in image_files[:10]:
    print(os.path.basename(img))
````

````python
image_dict = {
    os.path.basename(image_path): image_path
    for image_path in image_files
}

print("Images indexed:", len(image_dict))
````

````python
data["image_path"] = data["ID"].astype(str).map(image_dict)

print(data.head(10))
````

````python
print(
    "Matched images:",
    data["image_path"].notna().sum()
)

print(
    "Unmatched images:",
    data["image_path"].isna().sum()
)
````

````python
data = data.dropna(
    subset=["image_path"]
).reset_index(drop=True)

print("Final dataset shape:", data.shape)
````

````python
print(data["Class"].value_counts())
````

````python
fig, axes = plt.subplots(
    2, 5,
    figsize=(15, 6)
)

sample_data = data.sample(
    n=min(10, len(data)),
    random_state=42
)

for ax, (_, row) in zip(
    axes.ravel(),
    sample_data.iterrows()
):
    
    image = Image.open(
        row["image_path"]
    ).convert("RGB")
    
    ax.imshow(image)
    ax.set_title(row["Class"])
    ax.axis("off")

plt.tight_layout()
plt.show()
````

Part A Explanation

The dataset is downloaded using KaggleHub. The CSV file is loaded and each ID is matched directly with the corresponding image filename. Sample images are displayed with their age-group labels.

````python

## Part B: Preprocess the images

data["Class"] = (
    data["Class"]
    .astype(str)
    .str.strip()
    .str.upper()
)

print(data["Class"].value_counts())
````

````python
label_mapping = {
    "YOUNG": 0,
    "MIDDLE": 1,
    "OLD": 2
}

data["label"] = data["Class"].map(label_mapping)

print(
    data[["Class", "label"]].head(10)
)
````

````python
transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.5, 0.5, 0.5],
        std=[0.5, 0.5, 0.5]
    )
])
````

````python
class AgeDataset(Dataset):
    
    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.transform = transform
    
    def __len__(self):
        return len(self.dataframe)
    
    def __getitem__(self, index):
        
        row = self.dataframe.iloc[index]
        
        image = Image.open(
            row["image_path"]
        ).convert("RGB")
        
        label = int(row["label"])
        
        if self.transform:
            image = self.transform(image)
        
        return image, label
````

````python
dataset = AgeDataset(
    data,
    transform=transform
)

print("Dataset size:", len(dataset))
````

````python
image, label = dataset[0]

print("Image shape:", image.shape)
print("Label:", label)
````

Part B Explanation

Images are resized to 64×64, converted to tensors, and normalized. The three age groups are encoded as numerical labels 0, 1, and 2.

````python
## Part C: Split and batch

SEED = 42

torch.manual_seed(SEED)
random.seed(SEED)
np.random.seed(SEED)

train_size = int(
    0.7 * len(dataset)
)

val_size = len(dataset) - train_size

train_dataset, val_dataset = random_split(
    dataset,
    [train_size, val_size],
    generator=torch.Generator().manual_seed(SEED)
)

print("Training samples:", len(train_dataset))
print("Validation samples:", len(val_dataset))
````

````python
BATCH_SIZE = 32

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)
````

````python
images, labels = next(
    iter(train_loader)
)

print("Images shape:", images.shape)
print("Labels shape:", labels.shape)
````

Part C Explanation

The dataset is split into 70% training and 30% validation data. DataLoaders provide the images in batches of 32 for efficient CNN training.

````python
## Part D: Build the CNN
torch.manual_seed(42)

device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")

print("Using device:", device)
````

````python


class AgeCNN(nn.Module):
    
    def __init__(self):
        super().__init__()
        
        # =========================
        # Block 1
        # =========================
        self.block1 = nn.Sequential(
            
            nn.Conv2d(
                3, 32,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.BatchNorm2d(32),
            
            nn.Conv2d(
                32, 32,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.BatchNorm2d(32),
            
            nn.MaxPool2d(2)
        )
        
        # =========================
        # Block 2
        # =========================
        self.block2 = nn.Sequential(
            
            nn.Conv2d(
                32, 64,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.BatchNorm2d(64),
            
            nn.Conv2d(
                64, 64,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.BatchNorm2d(64),
            
            nn.MaxPool2d(2)
        )
        
        # =========================
        # Block 3
        # =========================
        self.block3 = nn.Sequential(
            
            nn.Conv2d(
                64, 128,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.BatchNorm2d(128),
            
            nn.Conv2d(
                128, 128,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.BatchNorm2d(128),
            
            nn.MaxPool2d(2)
        )
        
        # =========================
        # Block 4
        # =========================
        self.block4 = nn.Sequential(
            
            nn.Conv2d(
                128, 256,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.BatchNorm2d(256),
            
            nn.Conv2d(
                256, 256,
                kernel_size=3,
                padding=1
            ),
            nn.ReLU(),
            nn.BatchNorm2d(256),
            
            nn.MaxPool2d(2)
        )
        
        # =========================
        # Fully Connected
        # =========================
        self.fc = nn.Sequential(
            
            nn.Flatten(),
            
            nn.Linear(
                256 * 4 * 4,
                512
            ),
            nn.ReLU(),
            nn.Dropout(0.5),
            
            nn.Linear(
                512,
                128
            ),
            nn.ReLU(),
            nn.Dropout(0.3),
            
            nn.Linear(
                128,
                3
            )
        )
    
    
    def forward(self, x):
        
        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)
        x = self.block4(x)
        
        x = self.fc(x)
        
        return x
````

````python
model = AgeCNN().to(device)

print(model)
````

````python
images, labels = next(
    iter(train_loader)
)

images = images.to(device)

outputs = model(images)

print("Input:", images.shape)
print("Output:", outputs.shape)
````

Part D Explanation

The CNN contains four convolutional blocks. The filters increase from 32 to 64, 128, and 256. Each block contains two convolution layers, ReLU activation, Batch Normalization, and Max Pooling. Fully connected layers and dropout are used for classification. The final layer has three outputs for Young, Middle, and Old.

````python

## Part E: Train and evaluate

criterion = nn.CrossEntropyLoss()

optimizer = optim.Adam(
    model.parameters(),
    lr=0.001
)
````

````python
EPOCHS = 15

train_losses = []
val_losses = []

train_accuracies = []
val_accuracies = []

for epoch in range(EPOCHS):
    
    # ==================================
    # TRAINING
    # ==================================
    
    model.train()
    
    running_loss = 0.0
    correct = 0
    total = 0
    
    for images, labels in train_loader:
        
        images = images.to(device)
        labels = labels.to(device)
        
        # Clear old gradients
        optimizer.zero_grad()
        
        # Forward pass
        outputs = model(images)
        
        # Calculate loss
        loss = criterion(
            outputs,
            labels
        )
        
        # Backpropagation
        loss.backward()
        
        # Update weights
        optimizer.step()
        
        running_loss += (
            loss.item() *
            images.size(0)
        )
        
        _, predicted = torch.max(
            outputs,
            1
        )
        
        total += labels.size(0)
        
        correct += (
            predicted == labels
        ).sum().item()
    
    
    epoch_train_loss = (
        running_loss /
        len(train_loader.dataset)
    )
    
    epoch_train_accuracy = (
        correct / total
    )
    
    
    # ==================================
    # VALIDATION
    # ==================================
    
    model.eval()
    
    val_running_loss = 0.0
    val_correct = 0
    val_total = 0
    
    with torch.no_grad():
        
        for images, labels in val_loader:
            
            images = images.to(device)
            labels = labels.to(device)
            
            outputs = model(images)
            
            loss = criterion(
                outputs,
                labels
            )
            
            val_running_loss += (
                loss.item() *
                images.size(0)
            )
            
            _, predicted = torch.max(
                outputs,
                1
            )
            
            val_total += labels.size(0)
            
            val_correct += (
                predicted == labels
            ).sum().item()
    
    
    epoch_val_loss = (
        val_running_loss /
        len(val_loader.dataset)
    )
    
    epoch_val_accuracy = (
        val_correct /
        val_total
    )
    
    
    # ==================================
    # STORE RESULTS
    # ==================================
    
    train_losses.append(
        epoch_train_loss
    )
    
    val_losses.append(
        epoch_val_loss
    )
    
    train_accuracies.append(
        epoch_train_accuracy
    )
    
    val_accuracies.append(
        epoch_val_accuracy
    )
    
    
    print(
        f"Epoch [{epoch + 1}/{EPOCHS}] | "
        f"Train Loss: {epoch_train_loss:.4f} | "
        f"Train Acc: {epoch_train_accuracy:.4f} | "
        f"Val Loss: {epoch_val_loss:.4f} | "
        f"Val Acc: {epoch_val_accuracy:.4f}"
    )
````

````python
plt.figure(figsize=(8, 5))

plt.plot(
    train_losses,
    label="Training Loss"
)

plt.plot(
    val_losses,
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.title(
    "Training and Validation Loss"
)

plt.legend()
plt.grid()

plt.show()
````

````python
plt.figure(figsize=(8, 5))

plt.plot(
    train_accuracies,
    label="Training Accuracy"
)

plt.plot(
    val_accuracies,
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.title(
    "Training and Validation Accuracy"
)

plt.legend()
plt.grid()

plt.show()
````

````python
print(
    f"Final Training Accuracy: "
    f"{train_accuracies[-1] * 100:.2f}%"
)

print(
    f"Final Validation Accuracy: "
    f"{val_accuracies[-1] * 100:.2f}%"
)

print(
    f"Final Training Loss: "
    f"{train_losses[-1]:.4f}"
)

print(
    f"Final Validation Loss: "
    f"{val_losses[-1]:.4f}"
)
````

````python
class_names = [
    "Young",
    "Middle",
    "Old"
]

model.eval()

images, labels = next(
    iter(val_loader)
)

images = images.to(device)
labels = labels.to(device)

with torch.no_grad():
    
    outputs = model(images)
    
    predictions = torch.argmax(
        outputs,
        dim=1
    )

images = images.cpu()
labels = labels.cpu()
predictions = predictions.cpu()
````

````python
fig, axes = plt.subplots(
    2, 5,
    figsize=(15, 6)
)

for i, ax in enumerate(
    axes.ravel()
):
    
    image = images[i].permute(
        1, 2, 0
    )
    
    # Undo normalization
    image = image * 0.5 + 0.5
    
    image = torch.clamp(
        image,
        0,
        1
    )
    
    ax.imshow(image)
    
    actual = class_names[
        labels[i].item()
    ]
    
    predicted = class_names[
        predictions[i].item()
    ]
    
    ax.set_title(
        f"Actual: {actual}\n"
        f"Predicted: {predicted}"
    )
    
    ax.axis("off")

plt.tight_layout()
plt.show()
````
