# Version 3

```python
# Version 3


class Model(nn.Module):
  def __init__(self,num_features):
    super().__init__()

    self.network = nn.Sequential(
        nn.Linear(num_features,3), #index 0 (num_features=5, output=3)
        nn.ReLU() ,   #index 1
        nn.Linear(3,1),  #index 2 (input=3,output=1 in next layer)
        nn.Sigmoid()  #index 3

    )

  def forward(self,features):
    out = self.network(features)

    return out
```

```python
# Create dataset

features = torch.rand(10,5)
```

```python
features
```

```python
features.shape[1]
```

```python
#create model
model=Model(features.shape[1])
```

```python
#calling model for forward pass
#model.forward(features)
model(features) # better way
```

```python
# show model weights 
# model.linear.weight #for version one
#model.linear1.weight  #for version two
model.network[0].weight  #for version three and 0 and 2 number index need to get weight
```

# Extra

```python

# Base Class
class Rectangle:
    def __init__(self, length, width):
        print("Rectangle constructor called")
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


# Child Class (Square inherits Rectangle)
class Square(Rectangle):
    def __init__(self, side):
        print("Square constructor called")
        super().__init__(side, side)   # Calls Rectangle constructor

    def square_info(self):
        return f"Square area: {self.area()}"


# Child Class (Cube inherits Square)
class Cube(Square):
    def __init__(self, side):
        print("Cube constructor called")
        super().__init__(side)   # Calls Square constructor

    def volume(self):
        return self.length ** 3


# Create Object
cube = Cube(4)

print("Area:", cube.area())
print("Volume:", cube.volume())
```
