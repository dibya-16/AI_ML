```python
torch.sum(e,dim=0) #column wise
```

```python
torch.sum(e,dim=1) # row wise
```

```python
torch.mean(e)
```

```python
torch.mean(e,dim=0)
```

```python
torch.median(e)
```

```python
torch.max(e)
```

```python
torch.min(e)
```

```python
torch.prod(e) #multiply
```

```python
torch.var(e) #variance
```

```python
f
```

```python
g
```

```python
torch.matmul(f,g)
```

```python
#dot product
vector1=torch.tensor([1,2])
vector2=torch.tensor([3,4])
```
