# Custom Dataset

```python
from torch.utils.data import Dataset,DataLoader
```

```python
class CustomDataset(Dataset): 
    def __init__(self, features, labels): 
        self.features=features
        self.labels=labels

    def __len__(self): 
        return self.features.shape[0]

    def __getitem__(self,index): 
        return self.features[index], self.labels[index] #x= feature, y=label

        
```

```python
dataset= CustomDataset(X,y)
```

# Dataloader

```python
dataloader=DataLoader(dataset,batch_size=4,shuffle=False)# shuffle = true korle randomly split hobe 
```
