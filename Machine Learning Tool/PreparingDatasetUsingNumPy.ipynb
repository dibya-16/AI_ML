```python
import numpy as np

# dataset
# study hour, playtime, marks. study hour and playtime indepedent and feature. maarks dependent and target.
data = np.array([
    [9.5, 0.5, 95.5],
    [8.2, 1.2, 86.4],
    [7.8, 1.5, 82.1],
    [7.0, 1.0, 81.0],
    [6.5, 2.0, 74.5],
    [5.0, 1.5, 70.0],
    [4.5, 3.0, 60.5],
    [3.0, 2.5, 55.0],
    [2.0, 4.0, 44.0],
    [1.5, 4.5, 39.5]
])

data
```

```python
#randomizing the dataset

random_index=np.random.choice(10,size=(10,),replace=False) # 1st 10 is ending point for 0 to 9 index. replace=false.because same index double jaate nah ashe

print(random_index)

data=data[random_index]

print(data)
```

```python
#feature(x) & target(y) extraction

x=data[ : , :2] # row start to end shob nibe. column 0 and 1 nibe
y=data[ : ,2: ] # row start to end shob nibe. column 2 nibe
```

```python
# train, test & split

train_index= int(len(data)*0.8)  # 80% data nibo

x_train=x[ :train_index,:]
y_train=y[ :train_index,:]

x_test=x[train_index: ,]
y_test=y[train_index: ,]

print("x train")
print(x_train)

print("y train")

print(y_train)
print("x test")
print(x_test)
print("y test")
print(y_test)

```

```python
# doing this work more easily from scikit-learn.

from sklearn.model_selection import train_test_split


#feature(X) & target(Y) extraction

X=data[ : , :2]
Y=data[ : ,2: ]

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2) # means just 2 tah data nibe or we can say 20% data nibe

print("X train")
print(X_train)

print("Y train")

print(Y_train)
print("X test")
print(X_test)
print("Y test")
print(Y_test)

```
