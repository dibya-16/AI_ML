## Ndarray

```python
import numpy as np

# 1d array
array1 = np.array([1,2,3,4,5])


# 2d array
array2 = np.array([ [1,2,3], # row1
                  [4,5,6] #row2
                ])

#3d array

array3 = np.array([
    #1st floor
    [ [1,2,3], # row1
      [4,5,6]  ], #row2

    #2nd floor

     [[7,8,9], # row1
      [10,11,12]  ], #row2


    #3rd floor
      [[3,14,15], # row1
      [16,17,18]  ] #row2


])
```

## Array Attributes


```python
#dimesion
print(array1.ndim)
print(array2.ndim)
print(array3.ndim)

#shape
print(array1.shape) # print element for 1D array
print(array2.shape) # print (row size, column size) for 2D array
print(array3.shape) # print (floor size, row size, column size) for 3D array


#data type
print(array1.dtype)
print(array2.dtype)
print(array3.dtype)

#size
print(array1.size)
print(array2.size)
print(array3.size)





```

## Ndarray creation from existing data

```python
#list

lst = [10,20,30,40 , 40.5]

arr = np.array(lst, dtype=np.int32)

print(arr)
print(type(arr))
print(arr.dtype)


mixed_lst = [10, True , 'Hello' ] # mixed list ah convert hoy nah data type

arr = np.array(mixed_lst)

print(arr)
print(type(arr))
print(arr.dtype)


matrix = [ [1,2,3], # row1
           [4,5,6] #row2
                ]

arr = np.array(matrix)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
#tuple

tpl = (10,20,30)

arr = np.array(tpl,dtype=np.int32)
print(arr.dtype)
arr = arr.astype(np.int16)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
#set

st = { 1, 2, 3 }

arr = np.array(list(st))
# print(arr.dtype)
arr = arr.astype(np.int16)
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
# dictionary

dc = { 'a' : 10 , 'b' : 20 , 'c' :30 }

keys = dc.keys()
values = dc.values()
items = dc.items()
```

```python
#keys to ndarray

arr = np.array(list(keys))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#values to ndarray

arr = np.array(list(values))
print(arr)
arr = arr.astype(np.int16)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#items to ndarray

arr = np.array(list(items))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

## Indexing and Slicing

```python
#indexing

print(array1)
print(array2)

array1[2] = 100

print(array1)
array2[0][2]= 100 #(1st row, 3rd column)

print(array2)
```

```python
#slicing
#1D = arr[start:end:step]

array1_mod = array1[1:4]



array1_mod[2] = 200
print(array1_mod)
```

```python
#slicing
#1D = arr[start:end:step]
#copy

array1_mod = array1[1:4].copy()

array1[1] = 2


array1_mod[2] = 200
print(array1_mod)
print(array1)
```

```python
#2D slicing
# arr[ row_start : row_end : step , column_start : column_end : step ]

print(array2)

#getting a row

row_0 = array2[0:1 ,]

row_1= array2[1: , ]

print(row_0)
print(row_1)


#getting a column

col_0 = array2[:: ,0:1] # same col_0 = array2[ : ,0:1]
col_1 = array2[::,1:2]  # same col_0 = array2[ : ,1:2]

print(col_0)
print(col_1)

#getting a portion


portion = array2[:: , 0:2]

print(portion)
```

## Creating ndarray from scratch (Bonus)

```python
# np.zeros

arr = np.zeros((2,3),dtype=np.int8)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.zeros_like(array3) ## array3 shape same
print(arr)
print(arr.shape)
```

```python
#np.ones

arr = np.ones((3,4,3),dtype=np.int8)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)


arr = np.ones_like(array3) ## array3 shape same
print(arr)
print(arr.shape)
```

```python
#np.empty

arr = np.empty((4,3),dtype=np.int8)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)

arr = np.empty_like(array3) ## array3 shape same
print(arr)
print(arr.shape)
```

```python
#np.full

arr = np.full((4,3),np.inf)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)


arr = np.full_like(array3,np.inf,dtype=np.float64)
#arr=np.full_like(array3,10)

print(arr)
print(arr.shape)
print(arr.dtype)
```

## Creating matrix from linear algebra (Bonus)

```python
# diagonal matrix

diagonal_matrix = np.diag([1,2,3,4]) # diagonal wise values

print(diagonal_matrix)
print(diagonal_matrix.shape)
```

```python
# identity matrix

print("With only row value. here column value will be same as default")
identity_mat = np.eye(4)
print(identity_mat)
print(identity_mat.shape)


# np.eye(row , column,k)
print("Without k value")
mat = np.eye(3,4)
print(mat)

print("K value with -1")
mat = np.eye(3,4,-1) 
print(mat)

print("K value with 1")

mat = np.eye(3,4,1) 
print(mat)
```
