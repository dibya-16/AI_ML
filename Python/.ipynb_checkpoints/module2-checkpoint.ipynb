Max and Min Problem

```python
inp=input() # 5 7 11
numbers=inp.split()
x=int(numbers[0])
y=int(numbers[1])
z=int(numbers[2])
min=x
max=x

#min
if y<min:
    min=y
if z<min:
    min=z

#max
if y>max:
    max=y
if z>max:
    max=z
print(min, max)        
```

Even, Odd, Possitive and Negative problem

```python
n= int(input())
inp=input()
numbers=inp.split()
positive=0;negative=0;even=0;odd=0
for i in range(n):
    x=int(numbers[i])
    #positive or negative
    if x>0:
        positive+=1
    elif(x<0):
        negative+=1 
    # even or odd
    if(x%2!=0):
        odd+=1
    else:
        even+=1

print("Even number:",even)
print("Odd number:",odd)
print("Positive number:",positive)
print("Negative number:",negative)                   
```

Digits Problem

```python
t= int(input())
for i in range(t):
    numbers=int(input())
    if numbers==0:
        print(0)
        continue
    while numbers>0:
        print(numbers%10, end=" ")
        #numbers=int(numbers/10) and numbers//=10 same logic
        numbers//=10
    print()
```
