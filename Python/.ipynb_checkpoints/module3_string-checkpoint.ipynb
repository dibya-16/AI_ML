```python
# single line string
# string is immutable
prompt="What's your name?"
print(prompt)
print(type(prompt))
```

```python
# multi line string
messages="""Hello,
How are you?
How is your life going?"""
print(messages)
```

Indexing and Slicing

```python
string="How old are you?"
print(string[5])
```

```python
firstMessage= messages[0:24]
secondMessage= messages[0:24:2] # [start:end:step]
print(firstMessage)
#print(secondMessage)
```

```python
print(firstMessage[-1]) # space
print(firstMessage[-2]) # space
print(firstMessage[-3]) # w
print(firstMessage[-4]) # o
print(firstMessage[-5]) # H
```

```python
new_first_message=firstMessage[::-1]
print(new_first_message)
```

String Methods

Splitting, Joining and Formatted String

```python
#w3school theke string methopds gula shob dekhbo and practice korbo

string="Welcome to new ML Course, new ML course, hello ML."
processed_string=string.lower()
print(processed_string)
#length
print(len(string))
print("course" in processed_string)

# shurur kon index theke substring tah shuru hoise
start_index= processed_string.find("new")
print(start_index)
# shesher diker kon index theke substring tah shuru hoise
end_index=processed_string.rfind("new")
print(end_index)

# substring koybaar ache
count=processed_string.count("new")
print(count)
processed_string=string.replace("ML","AI/ML")
print(processed_string)

#string is immutable.so jodi change korte chao. notun ektay store korte hobe.main tah kokhono change hobe nah.dekho last ah change hoy nai string print korechi.
print(string)

```

Splitting, Joining and Formatted String

```python
#splitting, joining
prompt="What is ML ?"
tokens=prompt.split()
print(tokens)
print(type(tokens))
sentence="-".join(tokens)
print(sentence)
print(type(sentence))
```

```python
#formatted 
name="Dibya"
age=25
height=5.8123455
info=f"My name is {name}. I am {age} years old. My height is {height:.3} feet."# .3 means doshomik er por dui ghor jabe
print(info)
model_accuracy=0.834535
print(f"Model accuracy is {model_accuracy:.2%}") # :.2% .here .2 means doshomik er por 2 ghor jabe and % diye bujhaise 100 diye multiply hobe
```
