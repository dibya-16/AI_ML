Given a list of integers, extract all numbers that are divisible by 3 but not divisible by 5.

Input Format:
Space-separated integers

Output Format:
A list of numbers satisfying the condition

Example:
Input:
3 5 9 10 15 18 20 21

Output:
[3, 9, 18, 21]


```python
lst= input()
numbers=lst.split()
result=[]
for i in numbers:
    n=int(i)
    if(n%3==0 and n%5!=0):
        result.append(n)
print(result)        
        
```

Write a function compute(a, b) that:
Returns (a² + b²) / (a - b)
Use exception handling to manage:
Division by zero

Input Format:
Two values

Output Format:
Computed result OR error message


```python
numbers=input()
numbers=numbers.split()
a=int(numbers[0])
b= int(numbers[1])


try:
    def compute(a,b):
        return (a**2+ b**2) / (a - b)

    m= compute(a,b)    
except Exception:
    print("Error: Division not possible")
else:
    print(m)
    
    
```

Write a Python program that reads a text file and finds the most frequently occurring word.

Requirements:
Open and read data from a file (Example: data.txt)
Ignore case sensitivity ("Python" and "python" as the same)
Ignore punctuation marks
If multiple words have the same highest frequency, print any one of them

Input Format:
A text file containing multiple lines of text

Output Format:
The most frequent word

Example File Content (data.txt):
Python is easy to learn.
Python is powerful and Python is popular.

Output:
python


```python
with open("./data.txt") as file:
    text=file.read()
    text=text.lower()
    words=text.split()
    count={}
    

    for word in words:
        count[word]=count.get(word,0)+1

        

    most_frequently_occurring_word=""
    max_number=0
    for word in count:
        if(count[word]>max_number):
            max_number=count[word]
            most_frequently_occurring_word=word
            
        
    

        
    print(most_frequently_occurring_word)  
    
    
```

Write a program that converts a sentence into a list representing the length of each word.Ignore extra spaces

Input Format:

A single line sentence

OutputFormat:
A list of integers representing word lengths

Example:
Input:
Machine learning is fun

Output:
[7, 8, 2, 3]


```python
s= input()
words=s.split()
result= list(map(len, words))
print(result)
```

A number is called a Mirror Number if reversing its digits results in the same number.

Write a program to check whether a number is a mirror number.

Input Format:
An integer

Output Format:
"Mirror Number" or "Not Mirror Number"

Example:
Input:
121

Output:
Mirror Number


```python
number= int(input())
main_number=number
reverse_number=0
while number>0:
    value=number%10
    reverse_number=reverse_number*10+value
    number=number//10

if(main_number==reverse_number):
    print("Mirror Number")
else:
    print("Not Mirror Number")
```
