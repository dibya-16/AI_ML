Operators in Python

```python
x=13; y=3
sum=x+y
sub=x-y
multi=x*y
div=round(x/y,2)
rem=x%y

print(sum,sub,multi,div,rem)
```

```python
x=10;y=3
greater=x>y
greater_than_equal=x>=y
equal=x==y
not_equal=x!=y
print(greater,greater_than_equal,equal,not_equal)
```

```python
x=15;y=4;z=3
result=(x>y)or(y<z)
print(result)
```

Problem: Digit Summation

```python
input_value=input()
numbers=input_value.split()
x=int(numbers[0])
y=int(numbers[1])
last_digit_of_x=x%10
last_digit_of_y=y%10
sum=last_digit_of_x+last_digit_of_y
print(sum)
```
