```python
# list is mutable

# declaration 
numbers = [100,20,30,40,70,100,100] # int 
float_numbers = [10.20,3.15,2.4,2.7,3.5] # float 
fruits = ['apple', 'orange' , 'lichi'] # string 
mix = [10 , 10.20 , 'apple'] # mixed list

list_er_moddhe_list =[ [100,20,30] , [14,12,31] , [13,12,12] ]
print(type(numbers),type(float_numbers),type(fruits),type(mix))
print(type(numbers[3]))
```

```python
#access 

print(numbers[2])

numbers[2] = 90

print(numbers[2])

new_list = numbers[0:3]

print(new_list)
```

```python
#modification 

#adding a item 
numbers.append(80)
print(numbers)


#insertion at any position 
numbers.insert(0,45)
print(numbers)


#deleting the last value 
numbers.pop()

print(numbers)

#deleting a specific values 
numbers.remove(100)

print(numbers)
```

```python
numbers.sort()
print(numbers)

numbers.reverse()
print(numbers)
```

```python
numbers.sort(reverse=True)

print(numbers)
```

List as Queue

```python
# first in first out

queue = []

# insertion
queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)
queue.append(5)


#remove 
print(queue.pop(0))


# accessing the front element 

print(f"front e {queue[0]}")

queue.pop(0)

print(f"front e {queue[0]}")
```
