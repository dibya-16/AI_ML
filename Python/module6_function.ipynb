Arguments

```python
# positional arguments

def values(a, b, c):
    print(f"{a},{b},{c}")
values(10,20,30)    
```

```python
def values(*args):
    print(args)
values(10,20,30)    
```

```python
def values(*args):
    for i in args:
        print(i)
values(10,20,30) 
```

```python
# keyword arguments
def values(a, b, c):
    print(f"{a},{b},{c}")
values(a=10,c=30,b=20)

```

```python
def values(**kwards):
    print(kwards)
values(a=10,b=20,c=30)    
```

```python
def values(**kwards):
    for key,value in kwards.items():
        print(f"{key}:{value}")
values(a=20,c=50,b=25)        
```

Generator

```python
lst=[x for x in range(500)] #dataset and eta create korar jonno comprehension use korsi
#print(lst)

def data_loader(chunk_size,lst):
    for i in range(0,len(lst), chunk_size):
        yield lst[i:i+chunk_size] 
x=data_loader(5,lst)
print(next(x))
print(next(x))
```

Lambda, Map and Apply function

```python
import pandas as pd


df = pd.DataFrame({'temperature (F)': [32, 212, 98.6, 68]})

df
df['temperature (C)'] = df['temperature (F)'].apply(lambda x: (x - 32) * 5/9)
df
```

```python
import pandas as pd


df = pd.DataFrame({'A': [1, 2, 3, 4], 'B': [5, 6, 7, 8]})

df

df['A'] = df['A'].apply(lambda x: x ** 0.5)

print(df) # print df lekha or sudhu df lekha ekhane same.duita tei print hobe
```

```python
import pandas as pd

df = pd.DataFrame({'total_spent': [120, 600, 450, 800]})

# Creating a 'High Value' feature on the fly
df['is_high_value'] = df['total_spent'].apply(lambda x: 1 if x > 500 else 0)

df
```

```python
import pandas as pd
# apply and map kind of same kaaj e kore

df = pd.DataFrame({'total_spent': [120, 600, 450, 800]})

# Creating a 'High Value' feature on the fly
df['is_high_value'] = df['total_spent'].map(lambda x: 1 if x > 500 else 0)

df
```
