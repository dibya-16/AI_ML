Practice 1 :


Get the text from sample.txt , and count the following

Number of lines

Number of words

Number of characters

and save it to the counter_of_string.txt file

```python
from functools import reduce

with open("./sample_data/sample.txt","r") as file:
    string_list=file.readlines() # line count kore by readlines()
    print(string_list)
    #number of lines
    total_lines=len(string_list)
    print(total_lines)

    # total number of words

    number_of_words=list(map(lambda x: len(x.split()),string_list))
    print(number_of_words)
    total_number_of_words=reduce(lambda x,y: x+y, number_of_words )
    #print(total_number_of_words)

    # cleaning process

    # new line delete
    string_list= list(map(str.strip, string_list))
    #print(string_list)
    
    # space delete
    string_list=list(map(lambda x: x.replace(" ",""), string_list))
    #print(string_list)

    # koyta character

    number_of_characters=list(map(lambda x: len(x), string_list))
    #print(number_of_characters)

    # total character

    total_number_of_characters= reduce(lambda x,y: x+y, number_of_characters)

    #print(total_number_of_characters)

    with open("./sample_data/counter_of_string.txt", "w") as file:
        file.write(f"Total lines: {total_lines}\nTotal number of words: {total_number_of_words}\nTotal number of characters: {total_number_of_characters}")

    

    
    

    
```

Write then read with one open

```python
# w+ character write+read kore eksathe
# other character r+ and a+

with open("./sample_data/write_read.txt","w+") as file:
    file.write("Hello World")
    
    print(file.tell())
    file.seek(0)
    print(file.read())
    file.truncate(5) # 5 number index theke next shob delete korbo
    file.seek(0)
    print(file.read())
```
