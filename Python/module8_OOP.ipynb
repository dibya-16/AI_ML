Class & Objects

```python
#Base Class/ Parent Class

class Phone:
    category= "Electronics"

    #constructor

    def __init__(self,model,battery,camera,battery_percentage=100): # self= specific object
        self.model=model
        self.battery=battery
        self.camera=camera
        self.battery_percentage=battery_percentage
        
    #methods

    def charge(self,hour):
        print(f"Charge completed by {hour} hour")

    def capture(self,photo):
        if(self.battery_percentage<=0):
            print("No Charge")
        else:
            self.battery_percentage-=photo
            print(f"Photo Captured in {self.model}")


class Cooling_Mechanism:

    
    def __init__(self,cooling_method):
        
        self.cooling_method= cooling_method
        
    #methods
    
    def cooling_on(self):
        
        print(f"the system is being cold by {self.cooling_method}")
    
    

    

      
    
```

```python
#creating object

iphone=Phone("I-18",4500,45)
iphone.category="Super Electronics"
blueberry=Phone("B-18",4000,40)
samsung=Phone("S-18",3500,35)
samsung.model=3700
print(iphone.category)
print(blueberry.category) # class attribute
print(samsung.category)
print(iphone.model) # object attribute
print(samsung.model)

iphone.capture(5)
print(iphone.battery_percentage)
print(blueberry.battery_percentage)




```

Single Inheritance

```python
# Child class/ Derived Class

class SmartPhone(Phone):

    def __init__(self,model,battery,camera,processor):
        
        super().__init__(model,battery,camera)
        self.processor=processor

    #modified methods
    def charge(self,hour):

        print("Fast Charging in Process")
        super().charge(hour)
       



```

```python

pro=SmartPhone("G",5000,150,"Snapdragon")

print(pro.model)
pro.charge(5)
```

Multiple Inheritance 

```python
class SmartPhone_CoolingMode(SmartPhone, Cooling_Mechanism):
    def __init__(self, model,battery,camera,processor, cooling_method):
        SmartPhone.__init__(self, model,battery,camera,processor)
        Cooling_Mechanism.__init__(self,cooling_method)
        
```

```python
pro_cooling_phone=SmartPhone_CoolingMode("j",6000,70,"SD","Nitrogen")

print(pro_cooling_phone.processor) # SmartPhone class theke
print(pro_cooling_phone.battery) # Phone class theke
print(pro_cooling_phone.cooling_method) # Cooling_Mechanism class theke
print(pro_cooling_phone.cooling_on()) # Cooling_Mechanism class er method
print(pro_cooling_phone.charge(1)) # SmartPhone er modified charge jeta Phone class theke inherited


```

Polymorphism

```python
class Camera:
    def __init__(self,name):
        self.name=name

    #method
    def capture(self):
        print("A photo is captured")

        
```

```python
class Smart_Phone(Camera):

    def __init__(self,name,resolution):
        super().__init__(name)
        self.resolution=resolution

    #method overriding
    def capture(self):
        print("Photo is captured by SmartPhone")    

class DSLR(Camera):

    def __init__(self,name,resolution):
        super().__init__(name)
        self.resolution=resolution

    #method overriding
    def capture(self):
        print("Photo is captured by DSLR")    


class Drone(Camera):

    def __init__(self,name,resolution):
        super().__init__(name)
        self.resolution=resolution


    #method overriding
    def capture(self):
        print("Photo is captured by Drone") 

        
        
        
```

```python

phone=Smart_Phone("Phone", 50)
dslr= DSLR("DSLR",70)
drone=Drone("Drone", 40)

phone.capture()
dslr.capture()
drone.capture()
```

Encapsulation

```python
# design part handled by operator/ developer

class Mobile:

    def __init__(self,name,model,imei):
        self.__name=name # private
        self.__model=model # private
        self.__imei=imei # private

    def charge(self):
        print("Phone is charging")

    # method imei ta paite pari
    def imei_getter(self):
        return self.__imei


    def model_getter(self):
        return self.__model

    def name_getter(self):
        return self.__name

    def name_setter(self,name):
        self.__name=name

        


# outside world handled by user
#registration

apple=Mobile("IPhone", 18, "134xwrt7")

#print(apple.imei)

print(apple.imei_getter())

print(apple.model_getter())

print(apple.name_getter())

apple.name_setter("Iphone Pro")

print(apple.name_getter())



```
